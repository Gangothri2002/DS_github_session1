```python
import matplotlib.pyplot as plt
import numpy as np

n = len(cat)
plt.figure(figsize=(24, 10))

# Optional: Define a list of colors (can be extended)
colors = ['skyblue', 'salmon', 'lightgreen', 'orange', 'violet', 'gold', 'cyan', 'pink', 'gray']

for i in range(1, n):
    plt.subplot(3, 3, i)
    keys = visa_df[cat[i]].value_counts().keys()
    values = visa_df[cat[i]].value_counts().values
    
    # Bar plot with color
    bars = plt.bar(keys, values, color=colors[i % len(colors)])
    
    # Add numbers on top of each bar
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width() / 2, height, str(height),
                 ha='center', va='bottom', fontsize=10, color='black')
    
    plt.title(cat[i])
    plt.xticks(rotation=45)  # Optional: rotate x-axis labels for clarity
    plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

n = len(cat)
plt.figure(figsize=(28, 10))

# Optional: Define a list of colors (can be extended)
colors = ['skyblue', 'salmon', 'lightgreen', 'orange', 'violet', 'gold', 'cyan', 'pink', 'gray']

for i in range(1, n):
    plt.subplot(3, 3, i)
    keys = visa_df[cat[i]].value_counts(normalize=True).keys()
    values = visa_df[cat[i]].value_counts(normalize=True).values
    
    # Bar plot with color
    bars = plt.bar(keys, values, color=colors[i % len(colors)])
    
    # Add numbers on top of each bar
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width() / 2, height, str(height),
                 ha='center', va='bottom', fontsize=10, color='black')
    
    plt.title(cat[i])
    plt.xticks(rotation=45)  # Optional: rotate x-axis labels for clarity
    plt.tight_layout()

plt.show()
```
