```python
Q1 = wage_data.quantile(0.25)
Q3 = wage_data.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Create dataframes for outliers and non-outliers
outlier_data = visa_df[(wage_data < lower_bound) | (wage_data > upper_bound)]
non_outlier = visa_df[(wage_data >= lower_bound) & (wage_data <= upper_bound)]
```

```python
plt.figure(figsize=(12,9))
plt.subplot(3,2,1).hist(wage_data,bins=20)
plt.title('Histogram')
plt.subplot(3,2,2).boxplot(wage_data,vert=False)
plt.title('Box_plot')
plt.figure(figsize=(12,9))
plt.subplot(3,2,3).hist(outlier_data['prevailing_wage'],bins=20)
plt.title('Histogram with ourlier')
plt.subplot(3,2,4).boxplot(outlier_data['prevailing_wage'],vert=False)
plt.title('Boxplot with outlier')
plt.figure(figsize=(12,9))
plt.subplot(3,2,5).hist(non_outlier['prevailing_wage'],bins=20)
plt.title('Histogram without outlier')
plt.subplot(3,2,6).boxplot(non_outlier['prevailing_wage'],vert=False)
plt.title('Boxplot without outlier')
```
