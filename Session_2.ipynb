```python
dtypes=dict(visa_df.dtypes)
# iterate two i,j
# apply the condition
# if j=='object'
#append i in a list
# else
# append i in list
cat,num=[],[]
for i,j in dtypes.items():
    if j=='object':
        cat.append(i)
    else:
        num.append(i)

print(cat,num)
```

```python
dtypes=dict(visa_df.dtypes)
cat=[i for i,j in dtypes.items() if j=='object']
num=[j for i,j in dtypes.items() if j!='object']
print(cat,num)
```
