```python
names=['Ramesh','Suresh','Satish']
city=['Hyd','KMR','SEC']
Data=[names,city]
cols=['Names','City']
idx=['A','B','C']
pd.DataFrame(Data).T        
```

```python
names=['Ramesh','Suresh','Satish']
city=['Hyd','KMR','SEC']
Data=[names,city]
cols=['Names','City']
idx=['A','B','C']
pd.DataFrame(Data,index=cols,columns=idx).T   
```

```python
names=['Ramesh','Suresh','Satish']
city=['Hyd','KMR','SEC']
Data=zip(names,city)
cols=['Names','City']
idx=['A','B','C']
pd.DataFrame(Data,index=idx,columns=cols)
```

```python
names=['Ramesh','Suresh','Satish']
city=['Hyd','KMR','SEC']
Age=[20,30,40]
Data=zip(names,city,Age)

cols=['Names','City','Age']
idx=['A','B','C']
pd.DataFrame(Data,index=idx,columns=cols)
```

**Step-10: How to add a new data to existed dataframe**

```python
import pandas as pd
d={'Names':['Gango3','Chinni','Mummy'],'Age':[20,21,23],'City':['Hyd','KMR','SEC']}
df=pd.DataFrame(d)
df['Marks']=[100,200,300]    #adding a new column
df['Age']=[100,200,300]    #modify the exist column
df
```
