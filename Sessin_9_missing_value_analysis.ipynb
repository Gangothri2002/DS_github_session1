```python
dict1={'Name':['Ajay','Gango3','Moksha',np.nan],
       'Age':[20,24,np.nan,26],'City':[np.nan,'Hyd','KMR','DBK'],
      'Company':['Google','Microsoft','Apple','Meta']}
data1=pd.DataFrame(dict1)
data1
```

**fillna**

```python
data1.fillna(20)
```

```python
#firat select the column
# then apply fill na
data1['Name'].fillna('Manish')
```

**Draw back: Filling with Random value is not a Good approach**

- To avoid the we have some methods
    - backfill
    - bfill
    - ffill
    - pad

**pad**

```python
data1.fillna(method='pad')
```

```python
import warnings
warnings.filterwarnings('ignore')
data1.fillna(method='pad')
```

```python
import warnings
warnings.filterwarnings('ignore')
data1.fillna(method='pad',axis=1)
```

```python
import warnings
warnings.filterwarnings('ignore')
data1.backfill()
```

```python
import warnings
warnings.filterwarnings('ignore')
data1.bfill()
```

```python
import warnings
warnings.filterwarnings('ignore')
data1.ffill()
```

```python
import warnings
warnings.filterwarnings('ignore')
data1.pad()
```

**Mean-Median-Mode**

```python
#step-1: Select the column
#step-2: get the mean
#step-3: apply fill na

age_mean=data1['Age'].mean()
print('age_mean',age_mean)
data1['Age'].fillna(age_mean)
```

```python
city_mode=data1['City'].mode()
print('city_mode',city_mode)
data1['City'].fillna(city_mode)
```

```python
age_mean   #fixed value
```
